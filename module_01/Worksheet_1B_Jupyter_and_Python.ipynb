## 8. Checking AI-written code

An AI assistant wrote the function below to compute the average hours watched per subscriber. It runs without errors.

```python
def average_hours(subscribers):
    total = 0
    count = 0
    for subscriber in subscribers:
        if subscriber["hours_watched_last_30_days"]:
            total = total + subscriber["hours_watched_last_30_days"]
            count = count + 1
    return total / count

print(average_hours(subscribers))
```
