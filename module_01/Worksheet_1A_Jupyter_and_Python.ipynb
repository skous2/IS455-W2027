## 8. Checking AI-written code

An AI assistant wrote the function below to compute the average visits per member. It runs without errors.

```python
def average_visits(members):
    total = 0
    count = 0
    for member in members:
        if member["visits_last_30_days"]:
            total = total + member["visits_last_30_days"]
            count = count + 1
    return total / count

print(average_visits(members))
```
