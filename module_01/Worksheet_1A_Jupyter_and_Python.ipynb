## 5. Many members as a list of dictionaries

*McKinney 2.3 covers `for` loops and `if` statements, and McKinney 3.1 covers list comprehensions.*

**Worked example.** A `for` loop runs the same code once for each item in a list. A list comprehension builds a new list in one line, in the pattern `[what to keep for each item if condition]`.

````python
# Example: three customers in a list of dictionaries
customers = [
    {"name": "Ana", "cups": 3},
    {"name": "Ben", "cups": 0},
    {"name": "Cy",  "cups": 7},
]

for c in customers:              # c is one dictionary each time through the loop
    print(c["name"], c["cups"])

count = 0
for c in customers:
    if c["cups"] == 0:
        count = count + 1
print(count)

big_buyers = [c["name"] for c in customers if c["cups"] > 5]
print(big_buyers)
````

## 6. Functions

*McKinney 3.2, "Functions," covers writing your own functions.*

**Worked example.** `def` names a function and lists its input. `return` sends the answer back. `or` is `True` when either condition is `True`, and `and` needs both.

````python
# Example: a function that flags regular customers
def is_regular(customer):
    return customer["cups"] >= 5 or customer["member"]

print(is_regular({"cups": 7, "member": False}))   # expect True
print(is_regular({"cups": 1, "member": True}))    # expect True
print(is_regular({"cups": 1, "member": False}))   # expect False
````

## 8. Checking AI-written code

**Worked example.** Before you trust a function, run it on an input small enough to check by hand. If the output doesn't match your hand calculation, the function is wrong, even if it ran without an error.

````python
# Example: test a function on a case you can check by hand
def total_cups(customers):
    total = 0
    for c in customers:
        total = total + c["cups"]
    return total

print(total_cups([{"cups": 2}, {"cups": 3}]))   # by hand, 2 + 3 = 5
````

**Your turn.** An AI assistant wrote the function below to compute the average visits per member. It runs without errors.

````python
def average_visits(members):
    total = 0
    count = 0
    for member in members:
        if member["visits_last_30_days"]:
            total = total + member["visits_last_30_days"]
            count = count + 1
    return total / count

print(average_visits(members))
````
