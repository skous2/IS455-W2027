*McKinney 5.2, "Essential Functionality," covers this under "Indexing, Selection, and Filtering." McKinney 5.3 covers `.isin()`.*

**Worked example.** Single brackets with one name give one column. Double brackets give a DataFrame with the columns you list. A True/False condition inside the brackets keeps only the rows where it's `True`. Use `&` for *and*, `|` for *or*, and wrap each condition in parentheses.

## 5. Group and summarize

*McKinney 10.1 and 10.2 cover `groupby` and `agg`, and McKinney 10.5, "Pivot Tables and Cross-Tabulation," covers `pivot_table`.*

**Worked example.** `groupby` splits the rows into groups, computes something within each group, and returns one row per group. Inside `agg`, each named column is written as `new_name=("column", "function")`. `pivot_table` does the same with two groupings, one for the rows and one for the columns.

````python
display(shop.groupby("plan").agg(
    n=("customer", "count"),
    avg_cups=("cups", "mean"),
))
display(pd.pivot_table(shop, values="cups", index="member", columns="plan", aggfunc="mean"))
````

## 7. Merge two tables

*McKinney 8.2 covers this under "Database-Style DataFrame Joins."*

**Worked example.** `merge` attaches columns from a second table wherever the key column matches. `how="left"` keeps every row of the first table. After a merge, check that the row count didn't change and that the new columns have no unexpected blanks.

````python
discounts = pd.DataFrame({"plan": ["Gold", "Silver"], "discount": [0.10, 0.05]})
merged = shop.merge(discounts, on="plan", how="left")
display(merged)
print(len(shop), len(merged), merged["discount"].isna().sum())   # same row count, 0 blanks
````

## 9. Read a messier CSV file

*McKinney 6.1, "Reading and Writing Data in Text Format," lists the `read_csv` options.*

**Worked example.** Real exports rarely match `read_csv`'s defaults, so look at the raw text first and then tell pandas what it needs to know. `sep` names the character between columns, `skiprows` skips lines above the header, `dtype` keeps a column as text, `thousands` removes commas inside numbers, and `na_values` lists extra codes that mean missing. Two other options are useful on big files. `usecols` loads only the columns you list, and `nrows` loads only the first rows.

````python
import requests

url = "https://raw.githubusercontent.com/skous2/is455-w2027/main/data/exports/smoothie_menu_export.csv"
print(requests.get(url, timeout=10).text)     # look at the raw file first

menu = pd.read_csv(url, sep="|", skiprows=1, dtype={"item_code": str}, thousands=",", na_values=["-"])
display(menu)
menu.dtypes
````
